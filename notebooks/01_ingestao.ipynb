# Sprint 1 — coleta e integração do bruto

A coleta executável está em `../src/coleta_integracao.py`. Este notebook inspeciona a configuração e as evidências preservadas em `data/raw/`. O coletor recusa sobrescrever o bruto por padrão.

```python
from pathlib import Path
import json
import yaml
import pandas as pd

ROOT = Path.cwd()
if not (ROOT / 'config' / 'params.yaml').exists():
    ROOT = ROOT.parent

config = yaml.safe_load((ROOT / 'config' / 'params.yaml').read_text(encoding='utf-8'))
log = json.loads((ROOT / 'data' / 'raw' / 'coleta_integracao_log.json').read_text(encoding='utf-8'))
integrado = pd.read_csv(ROOT / 'data' / 'raw' / 'integrado.csv')
config['location'], log['integration'], integrado.head()
```
