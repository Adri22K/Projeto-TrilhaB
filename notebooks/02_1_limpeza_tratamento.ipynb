# Sprint 2.1 - Limpeza e tratamento

Este notebook inspeciona o bruto da Sprint 1, aplica as regras de limpeza e grava `data/interim/integrado_limpo.csv`.

```python
from pathlib import Path
import sys
import pandas as pd
import yaml

ROOT = Path.cwd()
if not (ROOT / 'data' / 'raw' / 'integrado.csv').exists():
    ROOT = ROOT.parent

sys.path.insert(0, str(ROOT / 'src'))
from prepara_sprint2 import inspect_raw, clean

config = yaml.safe_load((ROOT / 'config' / 'params.yaml').read_text(encoding='utf-8'))
raw = pd.read_csv(ROOT / 'data' / 'raw' / 'integrado.csv')
inspection = inspect_raw(raw, config['sprint2']['sentinels'])
cleaned, cleaning_log = clean(raw, config['sprint2']['sentinels'])

out_dir = ROOT / 'data' / 'interim'
out_dir.mkdir(parents=True, exist_ok=True)
cleaned.to_csv(out_dir / 'integrado_limpo.csv', index=False, encoding='utf-8')

inspection['rows'], cleaning_log
```
